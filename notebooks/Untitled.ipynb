```python
from azure.identity import DefaultAzureCredential
from azure.keyvault.secrets import SecretClient

def get_storage_connection_string():
    key_vault_url = "https://samirkey.vault.azure.net/"

    credential = DefaultAzureCredential()

    client = SecretClient(
        vault_url=key_vault_url,
        credential=credential
    )

    secret = client.get_secret("keystockagesamir")

    return secret.value
```

```python
from azure.storage.blob import BlobServiceClient
blob_service = BlobServiceClient.from_connection_string(
    get_storage_connection_string()
)

container = blob_service.get_container_client("raw")

for container_name in ["raw", "clean"]:
    print(f"\n--- {container_name.upper()} ---")

    container = blob_service.get_container_client(container_name)

    for blob in container.list_blobs():
        print(blob.name)
```
